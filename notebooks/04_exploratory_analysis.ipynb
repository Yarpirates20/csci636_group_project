```python
# Exploratory Analysis

#Setup
from pathlib import Path

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA = ROOT / "data"
```

```python
import pandas as pd

baseline_train = pd.read_csv(
    DATA / "processed" / "baseline" / "train.csv",
    parse_dates=["date"],
    index_col="date"
)

enhanced_train = pd.read_csv(
    DATA / "processed" / "enhanced" / "train.csv",
    parse_dates=["date"],
    index_col="date"
)
```

```python
print(baseline_train["target"].value_counts())
print(baseline_train["target"].value_counts(normalize=True))

print(enhanced_train["target"].value_counts())
print(enhanced_train["target"].value_counts(normalize=True))

display(baseline_train.describe())
display(enhanced_train.describe())

print(enhanced_train.isna().sum())

enhanced_train.hist(figsize=(15,12))
```
