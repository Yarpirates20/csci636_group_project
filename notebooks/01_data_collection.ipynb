```python
# Establish project root
from pathlib import Path

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA = ROOT / "data"

# print(ROOT)
```

```python
# Binance price data cleaning
import pandas as pd

columns = ["date", "open", "high", "low", "close"]

file = "../data/raw/binance_with_volume.csv"

price = pd.read_csv(file,header=0,usecols=range(5),names=columns)

# print(price.head())
```

```python
# Coinmetrics chain data cleaning

url = "https://raw.githubusercontent.com/coinmetrics/data/master/csv/btc.csv"

columns = [
    "time",
    "AdrActCnt",
    "TxCnt",
    "HashRate",
    "FeeTotNtv",
    "CapMVRVCur"
]

onchain = pd.read_csv(url, usecols=columns)

onchain = onchain.rename(columns={
    "time": "date",
    "AdrActCnt": "active_addresses",
    "TxCnt": "transaction_count",
    "HashRate": "hash_rate",
    "FeeTotNtv": "transaction_fees",
    "CapMVRVCur": "mvrv"
})

onchain = onchain[
    onchain["date"] >= "2017-12-01"
].copy()

print(onchain.isna().sum())

# print(onchain.head())
```

```python
# Save local copies of necessary data
# Save local copy
output_dir = Path("../data/raw")
output_dir.mkdir(parents=True, exist_ok=True)

price.to_csv(output_dir / "price_raw.csv", index=False)

onchain.to_csv(output_dir / "onchain_raw.csv", index=False)
```
